```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv('data.csv')
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
for col in df.columns:
    if df[col].nunique()==1:
        print(f"{col} has only one unique value")
```

```python
df['Attrition_flag']=df['Attrition'].map({'Yes':1,'No':0})
```

```python
print(df[['Attrition','Attrition_flag']].head())
print(df['Attrition_flag'].value_counts())
```

```python
df['AgeGroup']=pd.cut(df['Age'], bins=[18,25,35,45,55,65],
    labels=['18-25','26-35','36-45','46-55','56-65'])
```

```python
dept_attrition=df.groupby('Department')['Attrition_flag'].mean().sort_values(ascending=False)
print(dept_attrition)
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x=dept_attrition.index,y=dept_attrition.values)
plt.ylabel('Attition Rate')
plt.title('Attrition Rate by Department')
plt.xticks(rotation=0)
plt.show()
```

```python
overtime_attrition=df.groupby('OverTime')['Attrition_flag'].mean()
print(overtime_attrition)
```

```python
role_attrition=df.groupby('JobRole')['Attrition_flag'].mean().sort_values(ascending=False)
print(role_attrition)
```

```python
age_attrition=df.groupby('AgeGroup')['Attrition_flag'].mean()
print(age_attrition)
```

```python
numeric_df=df.select_dtypes(include=[np.number])
correlations=numeric_df.corr()['Attrition_flag'].sort_values(ascending=False)
print(correlations)
```

```python
df['OverTime_flag']=df['OverTime'].map({'Yes':1,'No':0})
numeric_df=df.select_dtypes(include=[np.number])
correlations = numeric_df.corr()['Attrition_flag'].sort_values(ascending=False)
print(correlations)
```

```python
plt.figure(figsize=(14,10))
sns.heatmap(numeric_df.corr(),cmap='coolwarm',center =0,annot=False)
plt.title('Correlation Heatmap')
plt.show()
```

```python
df['SalaryRange'] = pd.cut(df['MonthlyIncome'],
                            bins=[0, 3000, 6000, 10000, 20000],
                            labels=['Low (<3K)', 'Mid (3K-6K)', 'High (6K-10K)', 'Very High (10K+)'])
```

```python
salary_attrition = df.groupby('SalaryRange', observed=True)['Attrition_flag'].mean()
print(salary_attrition)
```
